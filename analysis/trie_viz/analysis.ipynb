```python
from pathlib import Path
import pandas as pd

from load import load_trie_dumps
from plot import plot_trie
```

## Load Trie Dumps
Point this to the directory containing the `trie_X_YYY_nodes.csv` files produced by `dump-trie-hcltj`.

```python
data_dir = Path("../../data/test_dump")
trie_dumps = load_trie_dumps(data_dir, full_only=False)
print(f"Loaded {len(trie_dumps)} tries: {', '.join(trie_dumps.keys())}")
```

## Visualize Tries
Draw the tree structure. Internal nodes are blue circles, leaf nodes are green squares.

```python
trie_name = "SPO"
if trie_name in trie_dumps:
    df = trie_dumps[trie_name]
    print(f"{trie_name}: {len(df)} nodes, max_depth={df['depth'].max()}")
    plot_trie(df, title=f"Trie {trie_name}");
```
